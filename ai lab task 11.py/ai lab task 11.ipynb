### Lab 10 - Data Pre-Processing

### Step 1: Importing Libraries

```python
import pandas as pd
import numpy as np
```

### Step 2: Reading the CSV File

```python
data = pd.read_csv('titanic.csv')
data.head()
```

### Step 3: Checking Shape (Rows and Columns)

```python
print('We have {} rows.'.format(data.shape[0]))
print('We have {} columns'.format(data.shape[1]))
```

### Step 4: Checking Null Values

```python
np.sum(pd.isnull(data))
```

### Step 5: Checking Unique Values of a Specific Column

```python
# 1. Pehle column banayein (Ye line sab se upar ho)
data['education'] = 'Higher Secondary'

# 2. Phir uski unique values check karein
print(data['education'].unique())

# 3. Poora table check karein
data.head()
```

### Step 6: Filling Null Values with Mode

```python
num = data['education'].mode()[0]
data['education'] = data['education'].fillna(num)

# verify null values are filled
print(np.sum(pd.isnull(data)))
```

### Step 7: Type Conversion to Int

```python
# Pehle text ki jagah number daalein (e.g., 12, 14, 16)
data['education'] = 12 

# Ab isay integer mein convert karein (Ab error nahi aaye ga)
data['education'] = data['education'].astype(int)

data.dtypes
data['education'] = data['education'].astype(np.int64)
data.dtypes
```

### Step 8: Dropping Unnecessary Columns

```python
# Agar 'id' nahi bhi hoga toh error nahi aaye ga
data.drop('id', axis=1, inplace=True, errors='ignore')
data.head()
```

### Step 9: Checking Datatypes of All Columns

```python
data.dtypes
```

### Step 10: Splitting Data into X and Y

```python
x = data.iloc[:, 0:-1]
print('Shape of x:', x.shape)

y = data.iloc[:, -1]
print('Shape of y:', y.shape)
```

### Step 11: Converting Object Columns to Int using Factorize

```python
cat_columns = x.select_dtypes(['object']).columns
x[cat_columns] = x[cat_columns].apply(lambda x: pd.factorize(x)[0])
x.dtypes
```
