### Check & Remove Duplicate Rows

```python
print('Duplicate Rows Found:', Dataset.duplicated().sum())

Dataset = Dataset.drop_duplicates()

print('Duplicate Rows After Removal:', Dataset.duplicated().sum())
```

### Value Counts – Survived Column

```python
if 'Survived' in Dataset.columns:
    print(Dataset['Survived'].value_counts())
```

### Group-By Analysis – Survival Rate by Gender

```python
if 'Sex' in Dataset.columns and 'Survived' in Dataset.columns:
    print(Dataset.groupby('Sex')['Survived'].mean())
```
